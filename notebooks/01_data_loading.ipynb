# 01. Data Loading

## Цель
Загрузить dataset, проверить его структуру, определить целевую переменную, выделить признаки для дальнейшего EDA и построения модели классификации бухгалтерских документов.

```python
import pandas as pd
import numpy as np
```

```python
activities_data_df = pd.read_excel("../data/activities_data.xlsx")
activities_data_df.head()
```

```python
invoices_classification_df = pd.read_excel("../data/invoices_classification.xlsx")
invoices_classification_df
```

```python
suppliers_data_df = pd.read_excel("../data/suppliers_data.xlsx")
suppliers_data_df
```

```python
dfs = {
    "activities_data": activities_data_df,
    "invoices_classification": invoices_classification_df,
    "suppliers_data": suppliers_data_df
}

for name, df in dfs.items():
    print("=" * 80)
    print(f"{name}")
    print(f"shape: {df.shape}")
    print("\ncolumns:")
    print(df.columns.tolist())
    print("\ndtypes:")
    print(df.dtypes)
    print("\nmissing values:")
    print(df.isna().sum().sort_values(ascending=False).head(20))
    print("\nhead:")
    display(df.head(3))
```

```python
for name, df in dfs.items():
    print("=" * 80)
    print(name)
    nunique_df = pd.DataFrame({
        "column": df.columns,
        "nunique": [df[col].nunique(dropna=False) for col in df.columns],
        "dtype": [df[col].dtype for col in df.columns]
    }).sort_values("nunique")
    display(nunique_df)
```

```python
for name, df in dfs.items():
    print("=" * 80)
    print(name)
    
    obj_cols = df.select_dtypes(include=["object"]).columns
    
    for col in obj_cols:
        nunique = df[col].nunique(dropna=False)
        if nunique <= 20:
            print(f"\n--- {col} ({nunique} unique) ---")
            print(df[col].value_counts(dropna=False).head(20))
```

```python
for name, df in dfs.items():
    print(name, "->", df.columns.tolist())
```

```python
merged_df = invoices_classification_df.merge(
    suppliers_data_df[["Supplier_Code", "Activity_Code", "Activity_Description"]],
    on="Supplier_Code",
    how="left"
)
merged_df.head()
```

```python
print("Shape after merge:", merged_df.shape)
print("\nMissing after merge:")
print(merged_df[["Supplier_Code", "Activity_Code", "Activity_Description"]].isna().sum())
```

```python
print("Количество классов:", merged_df["Account_Number"].nunique())
```

```python
account_map = (
    merged_df[["Account_Number", "Account_Description"]]
    .drop_duplicates()
    .sort_values("Account_Number")
)

account_map
```

```python
account_desc_check = (
    merged_df.groupby("Account_Number")["Account_Description"]
    .nunique()
    .sort_values(ascending=False)
)

account_desc_check
```

```python
problem_accounts = account_desc_check[account_desc_check > 1].index.tolist()
problem_accounts
```

```python
merged_df.loc[
    merged_df["Account_Number"].isin(problem_accounts),
    ["Account_Number", "Account_Description"]
].drop_duplicates().sort_values(["Account_Number", "Account_Description"])
```

## Предварительный вывод по целевой переменной

В качестве целевой переменной для задачи классификации будет использоваться `Account_Number`.

Поле `Account_Description` не является полностью консистентным: для некоторых кодов счета в датасете присутствуют разные текстовые описания. Поэтому `Account_Description` будет использоваться только для интерпретации результатов, но не как обучающая метка и не как признак модели.

```python
target_counts = merged_df["Account_Number"].value_counts()
print("Количество объектов:", len(merged_df))
print("Количество классов:", merged_df["Account_Number"].nunique())
print("Минимум объектов в классе:", target_counts.min())
print("Максимум объектов в классе:", target_counts.max())
print("Медиана объектов в классе:", target_counts.median())
target_counts
```

```python
merged_df.to_csv('../data/modified/merged_dataframe.csv', index=False)
```

## Итоги первого этапа

В ходе загрузки и первичного анализа данных были рассмотрены три таблицы:

- `activities_data` — справочник видов деятельности;
- `suppliers_data` — таблица поставщиков с привязкой к виду деятельности;
- `invoices_classification` — основная таблица с данными по строкам счетов и бухгалтерской классификацией.

Для формирования рабочей обучающей выборки таблица `invoices_classification` была объединена с `suppliers_data` по полю `Supplier_Code`, что позволило добавить информацию о виде деятельности поставщика.

В качестве целевой переменной выбрано поле `Account_Number`, так как оно представляет собой код бухгалтерского счета. Поле `Account_Description` оказалось неконсистентным: для некоторых кодов счета присутствуют разные текстовые описания. По этой причине `Account_Description` не будет использоваться ни как целевая переменная, ни как входной признак модели.

Следующим этапом будет выполнен разведочный анализ данных (EDA) для изучения распределения классов, текстовых признаков, числовых признаков и возможных ограничений набора данных.
